This notebook calls `recommendation_system.py` locally and store the results in this same file.

```python
from IPython.display import display, Markdown

from recommendation_system import CoordinatorAgent
from config import WEATHER_API_KEY, OPENAI_API_KEY

# Test dates 
test_cases = [ 
    ("2025-10-15", "Philippines"), 
    ("2025-10-16", "Philippines"), 
    ("2025-10-17", "Philippines"), 
    ("2025-10-18", "xAE"),              # Invalid location
    ("2025-12-21", "Philippines")       # No events on this date
] 
 
coordinator = CoordinatorAgent(WEATHER_API_KEY, OPENAI_API_KEY) 

# p.esguerra: using jupyter display so that output is not truncated
for test_date, test_location in test_cases: 
    display(Markdown(f"\n{'='*50}"))
    display(Markdown(f"## Getting recommendations for {test_location} on {test_date}:"))
    display(Markdown(f"{'='*50}"))
    display(Markdown(coordinator.get_recommendations(test_location, test_date)))
```
